```python
# 分组聚合
# df.groupby('分组的字段')['聚合字段'].聚合函数()
df = pd.read_csv('data/employees.csv')
df
```

```python
# 计算不同部门的平均薪资
df.groupby('部门').groups  #查看分组
df.groupby('部门').get_group('人事部')  #查看具体的某个分组数据
df2 = df.groupby('部门')[['月薪']].mean()
df2['月薪'] = df2['月薪'].round(2)
df2 = df2.reset_index()
df2.sort_values('月薪', ascending=False)
```

```python
# 计算不同城市不同部门的人的平均薪资
df3 = df.groupby(['城市', '部门'])[['月薪']].mean()
df3['月薪'] = df3['月薪'].round(2)
df3 = df3.reset_index()
df3.sort_values('月薪', ascending=False)
```

```python
df = pd.read_csv('data/penguins.csv')
# print(df.head(5))
df.info()
```

```python
#缺失值的检查
print(df.isna().sum())
# 查看所有缺失数据的行数据
df[df.isna().any(axis=1)]
# 去掉有缺失值的数据
df = df.dropna()
```

```python
df['sex'] = df['sex'].astype('category')
df['bill_ratio'] = df['bill_length_mm'] / df['bill_depth_mm']
```

```python
# 数据分箱-把体重分为三个等级
labels = ['低', '中', '高']
df['mass_level'] = pd.cut(df['body_mass_g'], bins=3, labels=labels)
print(df['mass_level'].value_counts())
```

```python
# 按岛屿、性别分组
df.groupby(['sex']).agg({
    'body_mass_g': ['mean', 'count'],
})

df.groupby(['island']).agg({
    'body_mass_g': ['mean', 'count'],
})

df.groupby(['sex', 'island']).agg({
    'body_mass_g': ['mean', 'count'],
})
```

```python
df = pd.read_csv('data/sleep_health_lifestyle.csv')
df.head()
df.info()
df.describe()
```

```python
# 数据清洗
df.isna().sum()
# print(df[df.isna().any(axis=1)])
df = df.drop(columns=['Sleep Disorder'])
df
```

```python
# 数据特征的构造
df['Gender'] = df['Gender'].astype('category')
df['Occupation'] = df['Occupation'].astype('category')
df['BMI Category'] = df['BMI Category'].astype('category')
df[['hign', 'low']] = df['Blood Pressure'].str.split('/', expand=True)
print(df.head())
```

```python
# 睡眠质量的分箱
labels = ['差', '中', '优']
df['Quality of level'] = pd.cut(df['Quality of Sleep'], bins=3, labels=labels)
age_labels = ['青少年', '中年', '老年']
df['age_level'] = pd.cut(df['Age'], bins=3, labels=age_labels)
df
```

```python
# 数据统计、分析
print(df['BMI Category'].value_counts())
```

```python
# 根据不同的bmi、年龄分组，睡眠质量
df.groupby(['BMI Category','Age']).agg({
    'Sleep Duration': 'mean',
    'Quality of Sleep': 'mean',
    'Stress Level': 'mean'
})
```
