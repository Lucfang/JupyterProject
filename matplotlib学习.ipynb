```python
import matplotlib.pyplot as plt
from matplotlib import rcParams

# rcParams['font.family'] = ['SimHei']
rcParams['font.sans-serif'] = ['SimHei']
```

```python
# 绘制折线图
#创建图表，设置大小
plt.figure(figsize=(10, 4))

#要绘图的数据
month = ['1月', '2月', '3月', '4月', '5月', '6月', '7月', '8月', '9月', '10月', '11月', '12月']
sales = [100, 159, 180, 400, 210, 340, 200, 100, 300, 200, 100, 800]
# 绘制折线图
plt.plot(month, sales, label='电子产品', color='orange', linewidth=1.2, linestyle='-', marker='o', markersize=3)
# 添加标题
plt.title('2025年销售趋势', color='red', fontsize=20)
# 添加坐标轴的标签
plt.xlabel('月份', fontsize=10)
plt.ylabel('销售额(万元)', fontsize=10)
#添加图例
plt.legend(loc='upper left', fontsize=10)
#添加网格线
# plt.grid(True)
plt.grid(axis='y', alpha=0.5, color='gray', linestyle='--')
#设置刻度字体大小
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=12)
#设置y轴的范围
plt.ylim(0, 900)
#在每个数据点上显示数值
for x, y in zip(month, sales):
    plt.text(x, y + 2, str(y), ha='center', va='bottom', fontsize=10, color='black')

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制柱状图

#创建图表，设置大小
plt.figure(figsize=(8, 4))
# 要绘图的数据
subjects = ['语文', '数学', '英语', '科学']
scores = [85, 92, 78, 88]

# 绘制柱状图
plt.bar(subjects, scores, label='小红', color='orange', width=0.2)

# 添加标题
plt.title('2025年成绩公布', color='red', fontsize=20)

# 添加坐标轴的标签
plt.xlabel('科目', fontsize=10)
plt.ylabel('分数', fontsize=10)

#添加图例
plt.legend(loc='upper left', fontsize=8)

#添加网格线
# plt.grid(True)
plt.grid(axis='y', alpha=0.5, color='gray', linestyle='--')

#设置刻度字体大小
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=12)

#设置y轴的范围
plt.ylim(0, 100)
#在每个数据点上显示数值
for x, y in zip(subjects, scores):
    plt.text(x, y + 2, str(y), ha='center', va='bottom', fontsize=10, color='black')

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制条形图

#创建图表，设置大小
plt.figure(figsize=(8, 3))
# 要绘图的数据
countries = ['US', 'China', 'Japan', 'Germany', 'India']
gdp = [80, 76, 74, 72, 60]

# 绘制条形图
plt.barh(countries, gdp, color='orange', height=0.5)

# 添加标题
plt.title('2025年GDP排名', color='red', fontsize=20)

#设置刻度字体大小
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=12)

#设置y轴的范围
plt.xlim(0, 100)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制饼图
#创建图表，设置大小
plt.figure(figsize=(10, 4))

#要绘图的数据
things = ['学习', '娱乐', '运动', '睡觉', '其他']
times = [6, 4, 1, 8, 5]
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff7799']  #配色
# 绘制饼图
plt.pie(times, labels=things,
        autopct='%1.1f%%',  #显示百分比
        startangle=90,  #调整初始化图的角度
        colors=colors  #设置饼图配色
        )
# 添加标题
plt.title('一日的时间分布', color='red', fontsize=20)
# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制圆环图
#创建图表，设置大小
plt.figure(figsize=(10, 4))

#要绘图的数据
things = ['学习', '娱乐', '运动', '睡觉', '其他']
times = [6, 4, 1, 8, 5]
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff7799']  #配色
# 绘制圆环图
plt.pie(times, labels=things,
        autopct='%1.1f%%',  #显示百分比
        startangle=90,  #调整初始化图的角度
        colors=colors,  #设置饼图配色
        wedgeprops={'width': 0.6},  #设置圆环的宽度
        pctdistance=0.6  #设置百分比的位置
        )
# 添加标题
plt.title('一日的时间分布', color='red', fontsize=20)
plt.text(0, 0, '总计100%', ha='center', va='bottom', fontsize=8, color='black')
# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制爆炸式饼图
#创建图表，设置大小
plt.figure(figsize=(10, 4))

#要绘图的数据
things = ['学习', '娱乐', '运动', '睡觉', '其他']
times = [6, 4, 1, 8, 5]
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff7799']  #
explode = [0.08, 0, 0, 0, 0]
# 绘制爆炸式饼图

plt.pie(times, labels=things,
        autopct='%1.1f%%',  #显示百分比
        startangle=0,  #调整初始化图的角度
        colors=colors,  #设置饼图配色
        explode=explode,  #设置突出块的位置
        shadow=True
        )
# 添加标题
plt.title('一日的时间分布', color='red', fontsize=20)
# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
import random

# 绘制散点图
#创建图表，设置大小
plt.figure(figsize=(10, 4))

#要绘图的数据
x = []
y = []
for i in range(500):
    temp = random.uniform(0, 10)
    x.append(temp)
    y.append(temp * 2 + random.gauss(0, 2))
# 绘制散点图
plt.scatter(x, y, label='数据', color='blue', alpha=0.5, s=20)
# 添加标题
plt.title('X变量与Y变量的关系', color='red', fontsize=20)
# 添加坐标轴的标签
plt.xlabel('X自变量', fontsize=10)
plt.ylabel('Y因变量', fontsize=10)
#添加图例
plt.legend(loc='upper left', fontsize=10)
#添加网格线
# plt.grid(True)
plt.grid(axis='y', alpha=0.5, color='gray', linestyle='--')
#设置刻度字体大小
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=12)
#设置y轴的范围
plt.ylim(0, 30)

plt.plot([0, 10], [0, 20], color='orange', linewidth=2)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

# 图形分析
1. 折线图：趋势随时间变化
2. 条形图/柱状图：类别之间对比
3. 饼图：整体组成比例
4. 散点图：两变量相关性
5. 箱线图：数据分布、异常

```python
# 多个图的绘制方法
plt.figure(figsize=(14, 8))

month = ['1月', '2月', '3月', '4月', '5月', '6月', '7月', '8月', '9月', '10月', '11月', '12月']
sales = [100, 159, 180, 400, 210, 340, 200, 100, 300, 200, 100, 800]

#动态图表的生成
f1 = plt.subplot(2, 2, 1)  #生成一个子图 行 列 索引
f1.plot(month, sales)

f2 = plt.subplot(222)
f2.bar(month, sales)

f3 = plt.subplot(2, 2, 3)
f3.scatter(month, sales)

f4 = plt.subplot(2, 2, 4)
f4.barh(month, sales)
```

```python
# 案例分析
# 1.导入库
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams

rcParams['font.sans-serif'] = ['SimHei']

# 2.导入数据
df = pd.read_csv('data/weather.csv')

# 绘制气温的趋势变化图
df['date'] = pd.to_datetime(df['date'])
df['mean_temp'] = (df['high_temp'] + df['low_temp']) / 2
plt.figure(figsize=(10, 10))
plt.plot(df['date'], df['high_temp'], label='最高气温')
plt.plot(df['date'], df['low_temp'], label='最低气温')
plt.plot(df['date'], df['mean_temp'], label='平均气温')
plt.title('2026年10月气温趋势变化图')
plt.xlabel('日期')
plt.ylabel('气温：℃')
plt.legend()
plt.show()


```
