```python
from IPython.lib.pretty import Printable

"""
题目1： 温度数据分析
某城市一周的最高温度(℃)为[28,30,29,31,32,30,29]
- 计算平均气温、最高气温和最低气温
- 找出气温超过30℃ 的天数
"""
```

```python
import numpy as np

temps = np.array([28, 30, 29, 31, 32, 30, 29])
print(temps)

print('平均气温：%.2f' % np.mean(temps))
print(f'最高气温:{np.max(temps)}')
print(f'最低气温:{np.min(temps)}')

print(f'气温超过30℃ 的天数：len(temps[temps > 30])')
print(np.where(temps > 30, 1, 0))
print(np.cumsum(np.where(temps > 30, 1, 0))[-1])
print(np.count_nonzero(np.where(temps > 30, 1, 0)))
```

```python
"""
题目2：学生成绩统计
某班级 5 名学生的数学成绩为[85,90,78,92,88]
- 计算成绩的平均分、中位数和标准差
- 将成绩转换为十分制(假设满分为10)
"""
```

```python
scores = np.array([85, 90, 78, 92, 88])
print(scores)
print(f'平均分:{np.mean(scores)}')
print(f'中位数:{np.median(scores)}')
print('标准差:%.3f' % np.std(scores))
print(scores / 10)
```

```python
"""
题目3：矩阵运算
给定矩阵 A = [[1,2],[3,4]] 和 B = [[5,6],[7,8]]
- 计算 A + B 和 A * B (逐元素乘法)
- 计算A 和 B的矩阵乘法(点积)
"""
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(A + B)
print(A * B)
print(A @ B)
print(np.dot(A, B))
```

```python
"""
题目4：随机数据生成
生成一个(3,4)的随机整数数组，范围[0,10)
- 计算每列的最大值和每行的最小值
- 将数组中的所有奇数替换为-1
"""
```

```python
np.random.seed(0)
arr = np.random.randint(0, 10, (3, 4))
print(arr)
print(f'每列的最大值:{np.max(arr, axis=0)}')  #axis=0代表列 axis=1代表行
print(f'每行的最小值:{np.min(arr, axis=1)}')
print(np.where(arr % 2 == 1, -1, arr))
arr[arr % 2 == 1] = -1
print(arr)
```

```python
"""
题目5：数组变形
创建一个1到12的一维数组，并转换为(3,4)的二维数组
- 计算每行的和与每列的平均值
- 将数组展平为一维数组
"""
arr = np.arange(1, 13)
print(arr)
arr = np.reshape(arr, (3, 4))
print(arr)
print(f'每行的和:{np.sum(arr, axis=1)}')
print(f'每列的平均值:{np.mean(arr, axis=0)}')
print(np.reshape(arr, 12))
```

```python
"""
题目6：布尔索引
生成一个(5,5)的随机数组，范围[0,20)
- 找出数组中大于10的元素
- 将所有大于10的元素替换为0
"""
```

```python
np.random.seed(0)
arr = np.random.randint(0, 20, (5, 5))
print(arr)
print(arr[arr > 10])
arr[arr > 10] = 0
print(arr)
```

```python
"""
题目7：统计函数应用
某公司6个月的销售额(万元)为[120,135,110,125,130,140]
- 计算销售额总和、均值和方差
- 找出销售额最高和最低的月份
"""
```

```python
arr = np.array([120, 135, 110, 125, 130, 140])
print(arr)
print(f'销售额总和:{np.sum(arr)}')
print('均值:%.3f' % np.mean(arr))
print('方差:%.3f' % np.var(arr))
print(f'销售额最高月份:{np.argmax(arr) + 1}')
print(f'销售额最低月份:{np.argmin(arr) + 1}')
```

```python
"""
题目8：数组拼接
给定A = [1,2,3] 和 B = [6,7,8]
- 将A和B水平拼接为一个新数组
- 将A和B垂直拼接为一个新数组
"""
```

```python
A = np.array([1, 2, 3])
B = np.array([6, 7, 8])
C = np.concatenate([A, B])
print(C)
print(np.reshape(C, (2, 3)))
```

```python
"""
题目9：唯一值与排序
给定数组 [2,1,2,3,1,4,3]
- 找出数组中的唯一值并排序
- 计算每个唯一值出现的次数
"""
```

```python
arr = np.array([2, 1, 2, 3, 1, 4, 3])
u_arr, counts = np.unique(arr, return_counts=True)
print(u_arr)
print(counts)
for i in range(len(u_arr)):
    print(u_arr[i])
    print(u_arr[i], ':', len(arr[arr == u_arr[i]]))
```

```python
"""
题目10：综合应用
某商店5天的销售额(万元)和成本(万元)如下：
销售额：[20,25,22,30,28]
成本：[15,18,16,22,20]
- 计算每天的利润(销售额-成本)
- 计算利润的平均值和标准差
- 找出利润最高的天数
"""
```

```python
money = np.array([20, 25, 22, 30, 28])
amt = np.array([15, 18, 16, 22, 20])
p = money - amt
print(f'每天的利润:{p}')
print(('利润的平均值:%.3f' % np.mean(p)))
print(('利润的标准差:%.3f' % np.std(p)))
print(f'利润最高的天数:{len(p[p == np.max(p)])}')
```
