# series的创建

```python
import numpy as np
import pandas as pd
```

```python
# series的创建
s = pd.Series([10, 2, 7, 8, 9, 6])
# 自定义索引
s = pd.Series([10, 2, 7, 8, 9, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])
# 定义name
s = pd.Series([10, 2, 7, 8, 9, 6], index=['A', 'B', 'C', 'D', 'E', 'F'], name='月份')
print(s)
```

```python
# 通过字典来创建
s = pd.Series({'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5, 'f': 6})
print(s)

s1 = pd.Series(s, index=['a', 'c'])
print(s1)
```

# series的属性
1. index:Series的索引对象
2. values:Series的值
3. dtype或dtypes:Series的元素类型
4. shape:Series的形状
5. ndim:Series的维度
6. size:Series的元素个数
7. name:Series的名称
8. loc[]:显示索引，按标签索引或切片
9. iloc[]:隐式索引，按位置索引或切片
10. at[]:使用标签访问单个元素
11. iat[]:使用位置访问单个元素

```python
# series属性
print(s.index)
print(s.values)
print(s.shape, s.ndim, s.size)
s.name = 'test'
print(s.dtype, s.dtypes, s.name)
print(s.loc['f'])  #显式索引
print(s.loc['c':'f'])
print(s.iloc[1])  #隐式索引
print(s.iloc[2:6])
print(s.at['f'])
print(s.iat[1])
```

# series数据访问

```python
# 访问数据
print(s['c'])
print(s[s > 3])
print(s.head(2))
print(s.tail(1))
```

# series案例

```python
"""
案例一：创建一个包含10名学生数学成绩的Series,成绩范围在50-100之间
计算平均分、最高分、最低分，并找出高于平均分的学生人数
"""
```

```python
np.random.seed(42)
scores = pd.Series(np.random.randint(50, 101, 10), index=['学生' + str(i) for i in range(1, 11)])
print(scores)
avg = scores.mean()
print(f'平均分:{avg}')
print(f'最高分:{scores.max()}')
print(f'最低分:{scores.min()}')
print(f'高于平均分的学生人数:{len(scores[scores > avg])}')
print(f'高于平均分的学生人数:{scores[scores > avg].count()}')
```

```python
"""
案例二：给定某城市一周每天的最高温度Series,完成以下任务：
- 找出温度超过30度的天数
- 计算平均温度
- 将温度从高到低排序
- 找出温度变化最大的两天
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33], index=['周一', '周二', '周三', '周四', '周五', '周六', '周日'])
"""
```

```python
temps = pd.Series([28, 31, 29, 32, 30, 27, 33], index=['周一', '周二', '周三', '周四', '周五', '周六', '周日'])
print(temps)
print(f'温度超过30度的天数:{temps[temps > 30].count()}')
print(f'平均温度:{temps.mean()}')
print(f'温度从高到低排序:')
t2 = temps.sort_values(ascending=False)
print(t2)
t3 = temps.diff().abs()  #计算series的变化值
print(f'温度变化最大的两：', *(t3.sort_values(ascending=False).keys()[:2].to_list()))
```

```python
"""
案例三：股票价格分析
给定某股票连续10个交易日的收盘价Series:
计算每日收益率(当日收盘价/前日收盘价 - 1)
找出收益率最高和最低的日期
计算波动率(收益率的标准差)
prices = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2023-01-01', periods=10))
"""
```

```python
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2],
                   index=pd.date_range('2026-11-01', periods=10))
print(prices)
profile = prices.pct_change()
print(f'每日收益率:')
print(profile)
print(f'收益率最高的日期:{profile.idxmax()}')
print(f'收益率最低的日期:{profile.idxmin()}')
print(f'波动率:{profile.std()}')
```

```python
"""
案例四：销售数据分析
某产品过去12个月的销售量Series:
计算季度平均销量(每3个月为一个季度)
找出销售最高的月份
计算月环比增长率
找出连续增长超过2个月的月份
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220],
                  index=pd.date_range('2022-01-01', periods=12, freq='MS'))
"""
```

```python
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220],
                  index=pd.date_range('2022-01-01', periods=12, freq='MS'))
print(sales)
print('季度平均销量(每3个月为一个季度):')
print(sales.resample('QS').mean()) #重新采样
print(f'销售最高的月份:{sales.idxmax()}')
print('月环比增长率:')
profile = sales.pct_change()
print(profile)
print('连续增长超过2个月的月份:')
t = profile > 0
t[t.rolling(3).sum() == 3].keys().tolist()
```

```python
"""
案例五：每小时销售数据分析
某商店每小时销售额Series:
按天重采样计算每日总销售额
计算每天营业时间(8:00-22:00)和非营业时间的销售额比例
找出销售额最高的3个小时

np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01', periods=24,freq='h'))
"""
```

```python
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01', periods=24,freq='h'))
print(hourly_sales)
day_sales = hourly_sales.resample('D').sum()
print(f'按天重采样计算每日总销售额:{day_sales}')
bussiness_hours_sales = hourly_sales.between_time('8:00','22:00').sum() #筛选一段时间内的series
print(f'每天营业时间(8:00-22:00)和非营业时间的销售额比例:{bussiness_hours_sales/(day_sales - bussiness_hours_sales)}')
# temp = hourly_sales[(hourly_sales.index.hour >= 8) & (hourly_sales.index.hour <= 22)]
# hourly_sales.drop(temp.index)
# mask = (hourly_sales.index.hour >= 8) & (hourly_sales.index.hour <= 22)
# b = hourly_sales[mask]
# n_b = hourly_sales[~mask]
# print(b.sum()/n_b.sum())
print(f'销售额最高的3个小时:{hourly_sales.nlargest(3).keys()}')
```
